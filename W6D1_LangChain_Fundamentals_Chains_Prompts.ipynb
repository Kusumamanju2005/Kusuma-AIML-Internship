```python
# Cell 9 — Import LangChain components

from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser

print("✅ LangChain components imported!")
```

```python
# Cell 10 — Create Ollama LLM

llm = ChatOllama(
    model="llama3.2:3b",
    temperature=0
)

print("✅ Ollama LLM created!")
```

```python
# Cell 11 — Create prompt template

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a helpful AI/ML mentor. "
        "Explain concepts clearly and simply for beginners."
    ),
    (
        "human",
        "{input}"
    )
])

print("✅ Prompt template created!")
```

```python
# Cell 12 — Output parser

output_parser = StrOutputParser()

print("✅ Output parser created!")
```

```python
# Cell 13 — Build LangChain chain

chain = prompt | llm | output_parser

print("✅ LangChain chain created!")
print("Flow: Prompt → Ollama LLM → Output Parser")
```

```python
# Cell 14 — Test chain

response = chain.invoke({
    "input": "What is Python?"
})

print("Response:")
print(response)
```

```python
# Cell 15 — Test chain with 5 inputs

questions = [
    "What is Python?",
    "What is machine learning?",
    "What is a database?",
    "What is an API?",
    "What is LangChain?"
]

for i, question in enumerate(questions, start=1):
    print(f"\n{'=' * 50}")
    print(f"Question {i}: {question}")
    print("=" * 50)

    answer = chain.invoke({
        "input": question
    })

    print(answer)
```

```python
# Cell 16 — Import ConversationBufferMemory

from langchain_classic.memory import ConversationBufferMemory
from langchain_classic.chains import LLMChain

print("✅ ConversationBufferMemory imported!")
```

```python
# Cell 17 — Create conversation memory

memory = ConversationBufferMemory(
    memory_key="history",
    return_messages=False
)

print("✅ Conversation memory created!")
```

```python
# Cell 18 — Prompt with conversation history

memory_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a helpful AI/ML mentor."
    ),
    (
        "human",
        """Conversation history:
{history}

User:
{input}

Answer the user's latest question while considering the conversation history."""
    )
])

print("✅ Memory prompt created!")
```

```python
# Cell 19 — Create memory chain

memory_chain = LLMChain(
    llm=llm,
    prompt=memory_prompt,
    memory=memory,
    verbose=False
)

print("✅ Memory chain created!")
```

```python
# Cell 20 — Conversation Turn 1

response = memory_chain.invoke({
    "input": "My name is Kusuma."
})

print(response["text"])
```

```python
# Cell 26 — Agent imports

from langchain_classic.agents import initialize_agent, AgentType
from langchain_core.tools import Tool

print("✅ Agent components imported!")
```

```python
# Cell 27 — Calculator function

def calculator(expression):
    try:
        allowed_chars = "0123456789+-*/(). %"

        if not all(char in allowed_chars for char in expression):
            return "Invalid mathematical expression."

        result = eval(
            expression,
            {"__builtins__": {}},
            {}
        )

        return str(result)

    except Exception as e:
        return f"Calculation error: {e}"

print("✅ Calculator created!")
```

```python
# Cell 28 — Test calculator

print("25 + 15 =", calculator("25 + 15"))
print("100 / 4 =", calculator("100 / 4"))
print("12 * 5 =", calculator("12 * 5"))
```

```python
# Cell 29 — Web search stub

def web_search(query):

    knowledge = {
        "python": "Python is a high-level programming language used for software development, automation, AI and data science.",

        "langchain": "LangChain is a framework for building applications powered by language models.",

        "machine learning": "Machine learning is a branch of AI where computers learn patterns from data."
    }

    query_lower = query.lower()

    for key, value in knowledge.items():
        if key in query_lower:
            return value

    return f"Search result for '{query}': No detailed information available in this demo."

print("✅ Web search stub created!")
```

```python
# Cell 30 — Test web search

print(web_search("What is Python?"))

print()

print(web_search("What is LangChain?"))
```

```python
# Cell 31 — Create LangChain tools

calculator_tool = Tool(
    name="Calculator",
    func=calculator,
    description=(
        "Useful for mathematical calculations. "
        "Input should be a mathematical expression."
    )
)

web_search_tool = Tool(
    name="WebSearch",
    func=web_search,
    description=(
        "Useful for searching simple information. "
        "Input should be a search query."
    )
)

tools = [
    calculator_tool,
    web_search_tool
]

print("Available tools:")

for tool in tools:
    print("✓", tool.name)
```

```python
# Cell 32 — Create LLM for agent

from langchain_ollama import OllamaLLM

agent_llm = OllamaLLM(
    model="llama3.2:3b",
    temperature=0
)

print("✅ Agent LLM created!")
```

```python
# Cell 33 — Create LangChain agent

agent = initialize_agent(
    tools=tools,
    llm=agent_llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True,
    handle_parsing_errors=True
)

print("✅ LangChain agent created!")
```

```python
# Cell 34 — Agent Task 1: Calculator

task1 = "Calculate 125 * 8 + 50"

result1 = agent.invoke({
    "input": task1
})

print("\nFinal Answer:")
print(result1["output"])
```

```python
# Cell 35 — Agent Task 2: Web Search

task2 = "Search for information about Python."

result2 = agent.invoke({
    "input": task2
})

print("\nFinal Answer:")
print(result2["output"])
```

```python
# Cell 36 — Agent Task 3: Combined task

task3 = "Tell me what LangChain is and calculate 20 * 5."

result3 = agent.invoke({
    "input": task3
})

print("\nFinal Answer:")
print(result3["output"])
```
