```python
import pandas as pd
import os
```

```python
data = {
    "Student_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Aarav", "Diya", "Rahul", "Priya", "Karan", "Sneha"],
    "State": ["Karnataka", "Karnataka", "Tamil Nadu", "Kerala", "Tamil Nadu", "Karnataka"],
    "Course": ["Python", "Python", "AI", "AI", "Python", "AI"],
    "Marks": [85, 90, 78, 88, 92, 81]
}

df = pd.DataFrame(data)
df.to_csv("indian_students.csv", index=False)

print(df)
```

```python
df = pd.read_csv("indian_students.csv")
```

```python
print("Shape:")
print(df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nFirst 10 Rows:")
print(df.head(10))
```

```python
# Students with marks greater than 85
filtered = df[df["Marks"] > 85]

print(filtered)
```

```python
# Average marks by state
grouped = df.groupby("State")["Marks"].mean()

print(grouped)
```

```python
courses = pd.DataFrame({
    "Course": ["Python", "AI"],
    "Duration": ["2 Months", "3 Months"]
})

merged = pd.merge(df, courses, on="Course")

print(merged)
```

```python
pivot = df.pivot_table(values="Marks",
                       index="State",
                       columns="Course",
                       aggfunc="mean")

print(pivot)
```
