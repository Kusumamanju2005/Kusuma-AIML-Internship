```python
# W5D1: First local inference using Llama 3.2

import requests

url = "http://127.0.0.1:11434/api/generate"

system_prompt = """
You are a helpful AI/ML mentor.
Explain technical concepts in simple language with examples.
"""

prompt = "What is Machine Learning? Explain it in simple words."

data = {
    "model": "llama3.2:3b",
    "system": system_prompt,
    "prompt": prompt,
    "stream": False
}

response = requests.post(url, json=data)

if response.status_code == 200:
    result = response.json()
    print("Prompt:", prompt)
    print("\nLlama 3.2 Response:\n")
    print(result["response"])
else:
    print("Error:", response.status_code)
    print(response.text)
```

```python
# W5D1: Test local LLM with 5 different prompts

import requests

url = "http://127.0.0.1:11434/api/generate"

system_prompt = """
You are a helpful AI/ML mentor.
Give clear, simple and accurate answers.
"""

prompts = [
    "What is Machine Learning?",
    "What is the difference between AI and ML?",
    "What is Python used for in AI?",
    "Explain overfitting in simple words.",
    "What is a neural network?"
]

for i, prompt in enumerate(prompts, 1):

    data = {
        "model": "llama3.2:3b",
        "system": system_prompt,
        "prompt": prompt,
        "stream": False
    }

    response = requests.post(url, json=data)

    print(f"\n{'='*60}")
    print(f"Prompt {i}: {prompt}")
    print(f"{'='*60}")

    if response.status_code == 200:
        result = response.json()
        print(result["response"])
    else:
        print("Error:", response.status_code)
```

```python
# W5D1: Compare Llama 3.2 and Qwen 2.5
# Same 3 questions are given to both models.

import requests

url = "http://127.0.0.1:11434/api/generate"

questions = [
    "What is Machine Learning? Explain in simple words.",
    "What is overfitting and how can we prevent it?",
    "What is the difference between AI and Generative AI?"
]

models = ["llama3.2:3b", "qwen2.5:3b"]

for question in questions:
    print("\n" + "=" * 80)
    print("QUESTION:", question)
    print("=" * 80)

    for model in models:
        data = {
            "model": model,
            "prompt": question,
            "stream": False
        }

        response = requests.post(url, json=data)

        print(f"\n--- {model} ---")

        if response.status_code == 200:
            result = response.json()
            print(result["response"])
        else:
            print("Error:", response.status_code)
```

```python
# W5D1: Llama 3.2 vs Qwen 2.5 Comparison Table

import requests
import pandas as pd
import time

url = "http://127.0.0.1:11434/api/generate"

questions = [
    "What is Machine Learning? Explain in simple words.",
    "What is overfitting and how can we prevent it?",
    "What is the difference between AI and Generative AI?"
]

models = ["llama3.2:3b", "qwen2.5:3b"]

results = []

for question in questions:
    for model in models:

        start_time = time.time()

        data = {
            "model": model,
            "prompt": question,
            "stream": False
        }

        response = requests.post(url, json=data)

        end_time = time.time()

        if response.status_code == 200:
            answer = response.json()["response"]
            response_time = round(end_time - start_time, 2)

            results.append({
                "Question": question,
                "Model": model,
                "Response Time (sec)": response_time,
                "Response": answer
            })

        else:
            results.append({
                "Question": question,
                "Model": model,
                "Response Time (sec)": None,
                "Response": "Error: " + response.text
            })

comparison_df = pd.DataFrame(results)

print("Llama 3.2 vs Qwen 2.5 Comparison")
print("=" * 80)

display(comparison_df)
```

```python
# W5D1: Save comparison results and document findings

# Save the comparison table as CSV
comparison_df.to_csv("W5D1_LLM_Comparison.csv", index=False)

print("Comparison results saved as: W5D1_LLM_Comparison.csv")

print("\n" + "=" * 70)
print("W5D1 FINAL CONCLUSION")
print("=" * 70)

print("""
I successfully installed Ollama and ran Llama 3.2 locally.
I tested the model with five different prompts using a custom
system prompt.

I also tested Llama 3.2 and Qwen 2.5 using the same three questions.
The responses were compared based on clarity, relevance, detail,
and response time.

Running LLMs locally provides privacy and reduces dependency on
external cloud APIs. However, local models require sufficient
computational resources.

Conclusion:
Both models produced useful responses, but their response style
and level of detail differed. Model selection should depend on
the required quality, speed, available hardware, and use case.
""")

print("\nFiles created:")
print("- W5D1_LLM_Comparison.csv")
```
