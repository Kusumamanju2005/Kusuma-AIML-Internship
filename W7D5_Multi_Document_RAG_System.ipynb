```python
import os
import shutil

for filename in os.listdir("/content"):
    if filename.endswith(".txt"):
        source = os.path.join("/content", filename)
        destination = os.path.join(documents_path, filename)

        shutil.move(source, destination)

print("Files inside W7D5 document folder:")

for filename in sorted(os.listdir(documents_path)):
    print("-", filename)
```

```python
from llama_index.core import SimpleDirectoryReader

documents = SimpleDirectoryReader(
    input_dir=documents_path
).load_data()

print("Number of documents loaded:", len(documents))
```

```python
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

embed_model = HuggingFaceEmbedding(
    model_name="BAAI/bge-small-en-v1.5"
)

print("Embedding model loaded successfully.")
```

```python
from llama_index.llms.ollama import Ollama

llm = Ollama(
    model="llama3.2:3b",
    request_timeout=120.0
)

print("Ollama LLM configured successfully.")
```

```python
from llama_index.core import Settings

Settings.llm = llm
Settings.embed_model = embed_model

print("LlamaIndex settings configured.")
```

```python
from llama_index.core import VectorStoreIndex

index = VectorStoreIndex.from_documents(
    documents
)

print("Multi-document RAG index created successfully.")
```

```python
query_engine = index.as_query_engine(
    similarity_top_k=3
)

print("Query engine created successfully.")
```

```python
query_engine = index.as_query_engine(
    similarity_top_k=3
)

print("Query engine recreated successfully.")
```

```python
response = query_engine.query(
    "What is Python and where is it commonly used?"
)

print("Answer:")
print(response)
```

```python
questions = [
    "What is machine learning?",
    "What is Retrieval Augmented Generation?",
    "What is LlamaIndex?",
    "What is a database?"
]

for question in questions:
    print("\n" + "=" * 60)
    print("Question:", question)

    response = query_engine.query(question)

    print("Answer:", response)
```

```python
def test_rag_query(question):
    response = query_engine.query(question)

    assert response is not None
    assert len(str(response).strip()) > 0

    return response


test_questions = [
    "What is Python?",
    "What is machine learning?",
    "What is RAG?"
]

for question in test_questions:
    answer = test_rag_query(question)
    print("PASS:", question)
    print("Answer:", answer)
    print()
```
