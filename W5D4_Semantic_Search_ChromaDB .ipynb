```python
import chromadb
import pandas as pd

from sentence_transformers import SentenceTransformer
from pypdf import PdfReader

print("Imports successful!")
print("ChromaDB version:", chromadb.__version__)
```

```python
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded!")
print("Embedding dimension:",
      len(embedding_model.encode("test sentence")))
```

```python
client = chromadb.Client()

collection = client.get_or_create_collection(
    name="w5d4_semantic_search",
    metadata={"hnsw:space": "cosine"}
)

print("Collection created:", collection.name)
```

```python
documents = [
    "Python is widely used for artificial intelligence and machine learning.",
    "Machine learning enables computers to learn patterns from data.",
    "Deep learning uses neural networks with multiple layers.",
    "Natural language processing helps computers understand human language.",
    "Computer vision allows machines to analyze images and videos.",
    "Data preprocessing prepares raw data for machine learning.",
    "Feature engineering creates useful features from raw datasets.",
    "Classification predicts discrete categories.",
    "Regression predicts continuous numerical values.",
    "Clustering groups similar data points.",
    "Embeddings represent text as numerical vectors.",
    "Vector databases store and search vector representations.",
    "Semantic search finds information based on meaning.",
    "Cosine similarity measures similarity between vectors.",
    "ChromaDB is a vector database for AI applications.",
    "FAISS provides efficient vector similarity search.",
    "Large language models generate text from prompts.",
    "RAG combines document retrieval with language generation.",
    "Ollama allows language models to run locally.",
    "Prompt engineering improves LLM responses."
]

metadata = [
    {"topic": "python"},
    {"topic": "machine_learning"},
    {"topic": "deep_learning"},
    {"topic": "nlp"},
    {"topic": "computer_vision"},
    {"topic": "preprocessing"},
    {"topic": "feature_engineering"},
    {"topic": "classification"},
    {"topic": "regression"},
    {"topic": "clustering"},
    {"topic": "embeddings"},
    {"topic": "vector_database"},
    {"topic": "semantic_search"},
    {"topic": "similarity"},
    {"topic": "chromadb"},
    {"topic": "faiss"},
    {"topic": "llm"},
    {"topic": "rag"},
    {"topic": "ollama"},
    {"topic": "prompt_engineering"}
]

print("Documents:", len(documents))
```

```python
embeddings = embedding_model.encode(
    documents,
    show_progress_bar=True
).tolist()

print("Embeddings generated!")
print("Number of embeddings:", len(embeddings))
print("Dimension:", len(embeddings[0]))
```

```python
ids = [f"doc_{i+1}" for i in range(len(documents))]

collection.add(
    ids=ids,
    documents=documents,
    embeddings=embeddings,
    metadatas=metadata
)

print("Documents added!")
print("Total documents:", collection.count())
```

```python
query = "How can computers understand human language?"

query_embedding = embedding_model.encode(query).tolist()

results = collection.query(
    query_embeddings=[query_embedding],
    n_results=5
)

print("Query:", query)
print("\nTop semantic search results:\n")

for i, document in enumerate(results["documents"][0]):
    print(f"{i+1}. {document}")
```

```python
print("Semantic Search Results:\n")

for i in range(len(results["documents"][0])):
    print(f"Rank {i+1}")
    print("Document:", results["documents"][0][i])
    print("Cosine distance:", results["distances"][0][i])
    print()
```

```python
filtered = collection.query(
    query_embeddings=[query_embedding],
    n_results=3,
    where={"topic": "nlp"}
)

print("NLP-filtered results:\n")

for i, document in enumerate(filtered["documents"][0]):
    print(f"{i+1}. {document}")
```

```python
print("MANUAL VERIFICATION")
print("=" * 50)

print("Query:", query)

for i, document in enumerate(results["documents"][0]):
    print(f"\nResult {i+1}:")
    print(document)

print("\nThe results are semantically related to the query.")
print("Smaller cosine distance indicates greater similarity.")
```
