```python
import chromadb

# Create an in-memory ChromaDB client
client = chromadb.Client()

# Create a collection
collection = client.create_collection(
    name="w6d4_documents",
    metadata={"hnsw:space": "cosine"}
)

print("ChromaDB client created successfully!")
print("Collection created:", collection.name)
```

```python
documents = [
    "Python is a popular programming language used for web development, automation, and data science.",
    "Machine learning allows computers to learn patterns from data and make predictions.",
    "Deep learning uses neural networks with multiple layers to learn complex patterns.",
    "Natural language processing helps computers understand and process human language.",
    "Computer vision enables computers to understand images and videos.",
    "Artificial intelligence focuses on creating systems that can perform tasks requiring human-like intelligence.",
    "A database is used to store, organize, and retrieve data efficiently.",
    "SQL is a language used to communicate with relational databases.",
    "MongoDB is a NoSQL database that stores data in flexible document-based structures.",
    "Git is a version control system used to track changes in software projects.",
    "GitHub is a platform used to host Git repositories and collaborate on code.",
    "APIs allow different software applications to communicate with each other.",
    "Flask is a lightweight Python framework used to build web applications.",
    "FastAPI is a modern Python framework for building high-performance APIs.",
    "ChromaDB is a vector database designed to store and search embeddings.",
    "Embeddings convert text or other data into numerical vectors.",
    "Cosine similarity measures the similarity between two vectors based on their direction.",
    "RAG stands for Retrieval-Augmented Generation and combines retrieval with language generation.",
    "LangChain provides tools for building applications powered by language models.",
    "Ollama allows developers to run large language models locally."
]

metadatas = [
    {"topic": "python"},
    {"topic": "machine_learning"},
    {"topic": "deep_learning"},
    {"topic": "nlp"},
    {"topic": "computer_vision"},
    {"topic": "ai"},
    {"topic": "database"},
    {"topic": "sql"},
    {"topic": "mongodb"},
    {"topic": "git"},
    {"topic": "github"},
    {"topic": "api"},
    {"topic": "flask"},
    {"topic": "fastapi"},
    {"topic": "chromadb"},
    {"topic": "embeddings"},
    {"topic": "similarity"},
    {"topic": "rag"},
    {"topic": "langchain"},
    {"topic": "ollama"}
]

ids = [f"doc_{i}" for i in range(1, 21)]

print("Number of documents:", len(documents))
print("Number of metadata entries:", len(metadatas))
print("Number of IDs:", len(ids))
```

```python
from chromadb.utils import embedding_functions

embedding_function = embedding_functions.DefaultEmbeddingFunction()

print("Embedding function created successfully!")
```

```python
# Delete the previous collection
try:
    client.delete_collection("w6d4_documents")
except Exception:
    pass

# Create collection with embedding function
collection = client.create_collection(
    name="w6d4_documents",
    embedding_function=embedding_function,
    metadata={"hnsw:space": "cosine"}
)

print("Collection recreated successfully!")
print("Collection name:", collection.name)
```

```python
collection.add(
    documents=documents,
    metadatas=metadatas,
    ids=ids
)

print("20 documents added successfully!")
print("Total documents in collection:", collection.count())
```

```python
query = "How can computers learn from data?"

results = collection.query(
    query_texts=[query],
    n_results=3
)

print("Query:", query)
print("\nTop 3 Similar Documents:\n")

for i, document in enumerate(results["documents"][0], 1):
    print(f"{i}. {document}")
```

```python
print("Query:", query)
print("\nSimilarity Search Results:\n")

for i in range(len(results["documents"][0])):
    document = results["documents"][0][i]
    distance = results["distances"][0][i]

    print(f"Result {i + 1}")
    print("Document:", document)
    print("Cosine distance:", distance)
    print("-" * 60)
```

```python
filtered_results = collection.query(
    query_texts=["How is data stored?"],
    n_results=5,
    where={"topic": "database"}
)

print("Query: How is data stored?")
print("Metadata filter: topic = database")
print("\nFiltered Results:\n")

for i, document in enumerate(filtered_results["documents"][0], 1):
    metadata = filtered_results["metadatas"][0][i - 1]

    print(f"{i}. {document}")
    print("   Metadata:", metadata)
```

```python
from reportlab.lib.pagesizes import A4
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer
from reportlab.lib.styles import getSampleStyleSheet

pdf_filename = "W6D4_RAG_Notes.pdf"

doc = SimpleDocTemplate(pdf_filename, pagesize=A4)
styles = getSampleStyleSheet()

content = [
    ("RAG Pipeline", "RAG stands for Retrieval-Augmented Generation. It combines information retrieval with a language model to generate answers using relevant external information."),
    ("ChromaDB", "ChromaDB is a vector database used to store embeddings and perform similarity searches."),
    ("Embeddings", "Embeddings represent text as numerical vectors. Similar pieces of text have vectors that are close to each other."),
    ("Cosine Similarity", "Cosine similarity measures the similarity between two vectors by comparing their direction."),
    ("LangChain", "LangChain is a framework for building applications using language models."),
    ("Ollama", "Ollama allows developers to run large language models locally."),
    ("RAG Process", "A typical RAG pipeline splits documents into chunks, creates embeddings, stores them in a vector database, retrieves relevant chunks, and sends them to a language model."),
    ("Benefits of RAG", "RAG helps language models answer questions using information from specific documents.")
]

story = []

for title, text in content:
    story.append(Paragraph(f"<b>{title}</b>", styles["Heading2"]))
    story.append(Paragraph(text, styles["BodyText"]))
    story.append(Spacer(1, 12))

doc.build(story)

print("PDF created successfully!")
print("File:", pdf_filename)
```

```python
from pypdf import PdfReader

reader = PdfReader(pdf_filename)

pdf_text = ""

for page in reader.pages:
    text = page.extract_text()
    if text:
        pdf_text += text + "\n"

print("PDF text extracted successfully!")
print("Number of pages:", len(reader.pages))
print("\nExtracted Text:\n")
print(pdf_text)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_text(pdf_text)

print("PDF split into chunks successfully!")
print("Number of chunks:", len(chunks))

print("\n===== CHUNKS =====\n")

for i, chunk in enumerate(chunks, 1):
    print(f"Chunk {i}:")
    print(chunk)
    print("-" * 60)
```

```python
# Create a separate collection for PDF chunks
try:
    client.delete_collection("w6d4_pdf_rag")
except Exception:
    pass

pdf_collection = client.create_collection(
    name="w6d4_pdf_rag",
    embedding_function=embedding_function,
    metadata={"hnsw:space": "cosine"}
)

print("PDF RAG collection created successfully!")
print("Collection:", pdf_collection.name)
```

```python
chunk_ids = [f"pdf_chunk_{i}" for i in range(len(chunks))]

chunk_metadatas = [
    {
        "source": pdf_filename,
        "chunk_id": i
    }
    for i in range(len(chunks))
]

pdf_collection.add(
    documents=chunks,
    metadatas=chunk_metadatas,
    ids=chunk_ids
)

print("PDF chunks added successfully!")
print("Total chunks stored:", pdf_collection.count())
```

```python
rag_query = "How does RAG work with ChromaDB and a language model?"

retrieved = pdf_collection.query(
    query_texts=[rag_query],
    n_results=3
)

print("RAG Query:")
print(rag_query)

print("\n===== TOP 3 RETRIEVED CHUNKS =====\n")

for i, chunk in enumerate(retrieved["documents"][0], 1):
    print(f"Chunk {i}:")
    print(chunk)
    print("-" * 70)
```

```python
retrieved_chunks = retrieved["documents"][0]

context = "\n\n".join(
    [f"Source Chunk {i+1}:\n{chunk}"
     for i, chunk in enumerate(retrieved_chunks)]
)

print("===== RETRIEVED CONTEXT =====")
print(context)
```

```python
from langchain_ollama import OllamaLLM

rag_llm = OllamaLLM(
    model="llama3.2:3b",
    temperature=0
)

rag_prompt = f"""
You are a helpful AI assistant.

Answer the user's question using ONLY the information provided in the context below.

Context:
{context}

Question:
{rag_query}

Answer:
"""

answer = rag_llm.invoke(rag_prompt)

print("===== RAG ANSWER =====")
print(answer)
```
