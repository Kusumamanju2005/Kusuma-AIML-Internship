```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
arr1 = np.array([10, 20, 30, 40, 50])

print(arr1)
print("Shape:", arr1.shape)
```

```python
arr2 = np.array([[1,2,3],
                 [4,5,6]])

print(arr2)
print("Shape:", arr2.shape)
```

```python
arr3 = np.array([
    [[1,2],[3,4]],
    [[5,6],[7,8]]
])

print(arr3)
print("Shape:", arr3.shape)
```

```python
a = np.array([1,2,3])

print(a + 5)
```

```python
x = np.array([10,20,30])
y = np.array([1,2,3])

print("Addition:", x+y)
print("Multiplication:", x*y)
```

```python
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print(np.matmul(A,B))
```

```python
data = {
    "Age":[20,21,22,23,24],
    "Marks":[70,75,80,90,95]
}

df = pd.DataFrame(data)

df.to_csv("sample_dataset.csv", index=False)

df
```

```python
df = pd.read_csv("sample_dataset.csv")

df
```

```python
print("Mean")
print(df.mean())

print("\nStandard Deviation")
print(df.std())

print("\nCorrelation")
print(df.corr())
```

```python
plt.plot(df["Age"], df["Marks"], marker="o")

plt.title("Age vs Marks")
plt.xlabel("Age")
plt.ylabel("Marks")

plt.grid(True)

plt.show()
```
