```python
# Cell 2 — Imports

import os
import numpy as np
import pandas as pd
import chromadb

from sentence_transformers import SentenceTransformer
from pypdf import PdfReader

print("Imports successful!")
print("ChromaDB version:", chromadb.__version__)
```

```python
# Cell 3 — Load Sentence Transformer

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")

# Test embedding
test_embedding = embedding_model.encode(
    "This is a test document."
)

print("Embedding dimension:", len(test_embedding))
```

```python
# Cell 4 — Create ChromaDB collection

client = chromadb.Client()

collection = client.get_or_create_collection(
    name="w5d3_documents",
    metadata={"hnsw:space": "cosine"}
)

print("Collection created successfully!")
print("Collection name:", collection.name)
```

```python
# Cell 5 — Create 20 sample documents

documents = [
    "Python is a popular programming language used for web development and artificial intelligence.",
    "Machine learning allows computers to learn patterns from data.",
    "Deep learning uses neural networks with multiple layers.",
    "Natural language processing helps computers understand human language.",
    "Computer vision enables machines to understand images and videos.",
    "Data preprocessing prepares raw data for machine learning models.",
    "Feature engineering creates useful features from raw data.",
    "Classification predicts categories or classes from input data.",
    "Regression predicts continuous numerical values.",
    "Clustering groups similar data points together.",
    "An embedding converts text into a numerical vector representation.",
    "Vector databases store and search high-dimensional vectors.",
    "Semantic search finds documents based on meaning rather than exact keywords.",
    "Cosine similarity measures the similarity between two vectors.",
    "ChromaDB is a vector database designed for AI applications.",
    "FAISS is a library designed for efficient similarity search.",
    "Large language models can generate human-like text from prompts.",
    "Retrieval augmented generation combines retrieval with language generation.",
    "Ollama allows users to run large language models locally.",
    "Prompt engineering improves the quality of responses from language models."
]

print("Number of documents:", len(documents))
```

```python
# Cell 6 — Create metadata

metadata = [
    {"topic": "python"},
    {"topic": "machine_learning"},
    {"topic": "deep_learning"},
    {"topic": "nlp"},
    {"topic": "computer_vision"},
    {"topic": "preprocessing"},
    {"topic": "feature_engineering"},
    {"topic": "classification"},
    {"topic": "regression"},
    {"topic": "clustering"},
    {"topic": "embeddings"},
    {"topic": "vector_database"},
    {"topic": "semantic_search"},
    {"topic": "similarity"},
    {"topic": "chromadb"},
    {"topic": "faiss"},
    {"topic": "llm"},
    {"topic": "rag"},
    {"topic": "ollama"},
    {"topic": "prompt_engineering"}
]

print("Metadata created:", len(metadata))
```

```python
# Cell 7 — Generate embeddings for 20 documents

embeddings = embedding_model.encode(
    documents,
    show_progress_bar=True
)

embeddings = embeddings.tolist()

print("Embeddings generated successfully!")
print("Number of embeddings:", len(embeddings))
print("Embedding dimension:", len(embeddings[0]))
```

```python
# Cell 8 — Add documents, embeddings and metadata

ids = [f"doc_{i+1}" for i in range(len(documents))]

collection.add(
    ids=ids,
    documents=documents,
    embeddings=embeddings,
    metadatas=metadata
)

print("Documents added successfully!")
print("Total documents in collection:", collection.count())
```

```python
# Cell 9 — Verify collection

stored_data = collection.get()

print("Total stored documents:", len(stored_data["ids"]))

print("\nFirst 5 documents:\n")

for i in range(min(5, len(stored_data["documents"]))):
    print(stored_data["ids"][i], "->", stored_data["documents"][i])
```

```python
# Cell 10 — Similarity search

query = "How do computers understand human language?"

query_embedding = embedding_model.encode(
    query
).tolist()

results = collection.query(
    query_embeddings=[query_embedding],
    n_results=5
)

print("Query:", query)

print("\nTop 5 Similar Documents:\n")

for i, document in enumerate(results["documents"][0]):
    print(f"{i+1}. {document}")
```

```python
# Cell 11 — Show similarity search results

print("Query:", query)
print("\nResults:\n")

for i in range(len(results["documents"][0])):
    print("Rank:", i + 1)
    print("Document:", results["documents"][0][i])
    print("Distance:", results["distances"][0][i])
    print("Metadata:", results["metadatas"][0][i])
    print("-" * 60)
```

```python
# Cell 12 — Metadata filtering

filtered_results = collection.get(
    where={"topic": "nlp"}
)

print("Documents with topic = NLP:\n")

for i, document in enumerate(filtered_results["documents"]):
    print(f"{i+1}. {document}")
```

```python
# Cell 13 — Similarity search with metadata filter

query = "How can computers process language?"

query_embedding = embedding_model.encode(
    query
).tolist()

filtered_search = collection.query(
    query_embeddings=[query_embedding],
    n_results=3,
    where={"topic": "nlp"}
)

print("Query:", query)

print("\nTop NLP Results:\n")

for i, document in enumerate(filtered_search["documents"][0]):
    print(f"{i+1}. {document}")
```

```python
# Cell 14 — Create a sample PDF for W5D3

!pip install -q reportlab

from reportlab.pdfgen import canvas

pdf_filename = "W5D3_Machine_Learning_Document.pdf"

c = canvas.Canvas(pdf_filename)

text = c.beginText(50, 750)
text.setFont("Helvetica", 11)

content = [
    "Machine Learning Overview",
    "",
    "Machine learning is a branch of artificial intelligence.",
    "It allows computers to learn patterns from data and make predictions.",
    "",
    "There are three common types of machine learning:",
    "1. Supervised learning",
    "2. Unsupervised learning",
    "3. Reinforcement learning",
    "",
    "Supervised learning uses labelled data for training.",
    "Classification and regression are common supervised learning tasks.",
    "",
    "Unsupervised learning works with data that does not have labels.",
    "Clustering is a common unsupervised learning technique.",
    "",
    "Feature engineering is an important step in machine learning.",
    "It involves creating useful features from raw data.",
    "",
    "Model evaluation helps determine how well a machine learning model performs.",
    "Accuracy, precision, recall and F1-score are commonly used metrics.",
    "",
    "Machine learning can be used in healthcare, finance, recommendation systems,",
    "computer vision, natural language processing and many other applications."
]

for line in content:
    text.textLine(line)

c.drawText(text)
c.save()

print("PDF created successfully!")
print("File:", pdf_filename)
```

```python
reader = PdfReader(pdf_filename)

pdf_text = ""

for page_number, page in enumerate(reader.pages):
    text = page.extract_text()

    if text:
        pdf_text += f"\nPage {page_number + 1}\n{text}"

print("PDF text extracted successfully!")
print("Number of characters:", len(pdf_text))

print("\nPreview:\n")
print(pdf_text[:2000])
```

```python
# Cell 16 — Split PDF text into chunks

chunk_size = 800
overlap = 100

chunks = []

start = 0

while start < len(pdf_text):
    end = start + chunk_size

    chunk = pdf_text[start:end].strip()

    if chunk:
        chunks.append(chunk)

    start += chunk_size - overlap

print("Number of chunks:", len(chunks))

print("\nFirst chunk:\n")
print(chunks[0][:1000])
```

```python
# Cell 17 — Limit chunks if PDF is very large

max_chunks = 100

chunks = chunks[:max_chunks]

print("Chunks used for embedding:", len(chunks))
```

```python
# Cell 18 — Generate embeddings for PDF chunks

pdf_embeddings = embedding_model.encode(
    chunks,
    show_progress_bar=True
)

pdf_embeddings = pdf_embeddings.tolist()

print("PDF embeddings generated!")
print("Number of embeddings:", len(pdf_embeddings))
print("Embedding dimension:", len(pdf_embeddings[0]))
```

```python
# Cell 19 — Create a separate ChromaDB collection for PDF

pdf_collection = client.get_or_create_collection(
    name="w5d3_pdf_collection",
    metadata={"hnsw:space": "cosine"}
)

print("PDF collection created!")
```

```python
# Cell 20 — Store PDF chunks

pdf_ids = [
    f"pdf_chunk_{i+1}"
    for i in range(len(chunks))
]

pdf_metadata = [
    {
        "source": pdf_filename,
        "chunk": i + 1
    }
    for i in range(len(chunks))
]

pdf_collection.add(
    ids=pdf_ids,
    documents=chunks,
    embeddings=pdf_embeddings,
    metadatas=pdf_metadata
)

print("PDF chunks stored successfully!")
print("Total PDF chunks:", pdf_collection.count())
```

```python
# Cell 21 — Retrieve top 3 PDF chunks

question = "What is the main topic discussed in this document?"

question_embedding = embedding_model.encode(
    question
).tolist()

pdf_results = pdf_collection.query(
    query_embeddings=[question_embedding],
    n_results=3
)

print("Question:", question)

print("\nTop 3 Retrieved Chunks:\n")

for i, chunk in enumerate(pdf_results["documents"][0]):
    print(f"\n--- Chunk {i+1} ---")
    print(chunk)
```
