```python
import requests

OLLAMA_URL = "http://127.0.0.1:11434/api/chat"
MODEL = "llama3.2:3b"

system_prompt = """
You are a helpful AI/ML mentor.
Explain technical concepts in simple language suitable for a beginner.
Keep your answers clear, practical, and concise.
When giving examples, use Python where appropriate.
"""

print("Model:", MODEL)
print("\nCustom System Prompt:")
print(system_prompt)
```

```python
import requests

def ask_ollama(prompt):
    payload = {
        "model": MODEL,
        "messages": [
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        "stream": False
    }

    response = requests.post(OLLAMA_URL, json=payload)
    response.raise_for_status()

    return response.json()["message"]["content"]

print("Ollama API function created successfully.")
```

```python
test_prompts = [
    "What is Machine Learning?",
    "Explain Python lists with a simple example.",
    "What is the difference between AI and Machine Learning?",
    "Explain overfitting in simple terms.",
    "What is a neural network?"
]

for i, prompt in enumerate(test_prompts, 1):
    print(f"\n{'='*60}")
    print(f"Prompt {i}: {prompt}")
    print(f"{'='*60}")

    try:
        answer = ask_ollama(prompt)
        print(answer)
    except Exception as e:
        print("Error:", e)
```

```python
import requests

OLLAMA_URL = "http://127.0.0.1:11434/api/chat"

comparison_questions = [
    "Explain Machine Learning in simple terms.",
    "What is the difference between a list and a tuple in Python?",
    "Explain overfitting with a simple real-world example."
]

def ask_model(model, prompt):
    payload = {
        "model": model,
        "messages": [
            {
                "role": "user",
                "content": prompt
            }
        ],
        "stream": False
    }

    response = requests.post(OLLAMA_URL, json=payload)
    response.raise_for_status()

    return response.json()["message"]["content"]


for i, question in enumerate(comparison_questions, 1):
    print("\n" + "=" * 70)
    print(f"QUESTION {i}: {question}")
    print("=" * 70)

    print("\n--- Llama 3.2:3b ---")
    print(ask_model("llama3.2:3b", question))

    print("\n--- Qwen 2.5:3b ---")
    print(ask_model("qwen2.5:3b", question))
```
